## prep other vars

```python
# related to variables potentially influencing phenotype definition
df3[['AGE']] = df3[['AGE']] #SG_UF_NOT','CS_ESCOL_N','EVOLUCAO','HOSPITALIZ','DT_INTERNA', 'CLASSI_FIN',     
df3['SEXO']=df3['CS_SEXO'].astype('category')
df3['GESTANT']=[1 if x in ['pregnant - timeline unknwon','2nd trimester','1st trimester','3rd trimester'] else 0 for x in df3['CS_GESTANT']]
df3['RACA']=df3['CS_RACA'].astype('category')
df3['Dengue']=[1 if x in ['Dengue', 'Alarm signs dengue','severe Dengue'] else 0 for x in df3['CLASSI_FIN']]
df3['SOROTIPO']=df3['SOROTIPO'].astype('category')
df3['NewStates']=[1 if x in ['42','43'] else 0 for x in df3['SG_UF_NOT']]
# possibly influenced by phenotypes
df3['HOSPITALIZ']=df3['HOSPITALIZ'].astype('category')
df3['DT_INTERNA']=df3['DT_INTERNA'].astype('category')
```

## prepare function

```python
import os
import matplotlib.pyplot as plt
def exploreCovariateAssoc(predicted, data, path,
                          varTestChi,
                           phenotype_models_str,
                          hosp_models_str,
                           severe_models_str,
                           severe_curv_models_str
                         ):
    data['predicted']=[f'cl_{x}' for x in predicted]
    data['predicted']= data['predicted'].astype('category')    
    

    if path is not None:
            if not os.path.exists(path):
                os.makedirs(path)
            if not os.path.exists(f'{path}/testChi'):
                os.makedirs(f'{path}/testChi')
            if not os.path.exists(f'{path}/MNLogit'):
                os.makedirs(f'{path}/MNLogit')
            if not os.path.exists(f'{path}/ProbitM_hosp'):
                os.makedirs(f'{path}/ProbitM_hosp')
            if not os.path.exists(f'{path}/ProbitM_severe'):
                os.makedirs(f'{path}/ProbitM_severe')
            if not os.path.exists(f'{path}/ProbitM_sevCurv'):
                os.makedirs(f'{path}/ProbitM_sevCurv')
    
    fig, ax = plt.subplots(figsize=(12,6))
    for label, df in data.groupby('predicted'):
        ax=df.AGE.plot(kind="kde", ax=ax, label=label)
    plt.legend()
    plt.xlim(0,15)
    plt.savefig(f'{path}/age_classes.png')
    plt.show()
    print(f'============================ Chi-squared tests ==========================')
    blind_chisq(data=data,covariates=varTestChi,predictedVal='predicted', path=f'{path}/testChi')
    print(f'============================ Multinomial logit with nested models ==========================')
    comps_df, pardict2, fits=compareModels_MNLogit(models_str=phenotype_models_str,data=data,path=f'{path}/MNLogit')
    
    display(comps_df)
    display(pardict2)
    print(f'============================ Hospitalization: Probit model with nested models ==========================')

    res_hosp, fits_hosp=compareProbitModels(hosp_models_str,data=data,path=f'{path}/ProbitM_hosp')
    
    display(res_hosp)
    print(f'============================ Severoty: Probit model with nested models ==========================')
    res_severe, fits_severe=compareProbitModels(severe_models_str,data=data,path=f'{path}/ProbitM_severe')
    
    display(res_severe)
    print(f'============================ severity curvilinear: Probit model with nested models ==========================')
    res_severeCurv, fits_severeCurv=compareProbitModels(severe_curv_models_str,data=data,path=f'{path}/ProbitM_sevCurv')
    
    display(res_severeCurv)

    res=[{'multinomial_pheotype':{'comparisons':comps_df,'parameters':pardict2,'fits':fits},
          'hosp_probit':{'results':res_hosp,'fits':fits_hosp},
          'severe':{'results':res_severe, 'fits':fits_severe},
          'severeCurvilinear':{'results':res_severeCurv, 'fits':fits_severeCurv}}]
          

    return res
```
